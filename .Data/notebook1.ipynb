```python
import pandas as pd

```

```python
df = pd.read_csv('World-happiness-report-updated_2024.csv', encoding='latin1')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum
```

```python
df[df.isnull().any(axis=1)]
```

```python
numeric_cols= [
    'Life Ladder', 
    'Log GDP per capita', 
    'Social support', 
    'Healthy life expectancy at birth', 
    'Freedom to make life choices', 
    'Generosity', 
    'Perceptions of corruption', 
    'Positive affect', 
    'Negative affect'
]

#filling the missing areas with median values 

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())


print(df[numeric_cols].isnull().sum)
```

```python
df.head(200)
```

```python
#rename the columns for better readability

df.rename(columns={
    'Log GDP per capita' : 'GDP',
    'Country name' : 'Country',
    'Healthy life expectancy at birth' : 'Life Expectancy',
    'Perceptions of corruption' : 'Corruption',
    'Freedom to make life choices': 'Freedom'

},
          inplace=True)


df.head(200)
```

```python
#sorting by the year 

df['year'] = pd.to_numeric(df['year'], errors='coerce')
df_sorted = df.sort_values(by='year', ascending=False)
df_sorted = df_sorted.reset_index(drop=True) #removes the index where it shows the countries list position in the database

df_sorted.head()
```

```python
import matplotlib.pyplot as plt

#show the top 10 by the 'Positive affect' 

top10 = df.sort_values(by='Positive affect', ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top10['Country'], top10['Positive affect'], color="green")
plt.xlabel('Positive affect')
plt.title('Top 10 countries with positive affect')
plt.show()
```
